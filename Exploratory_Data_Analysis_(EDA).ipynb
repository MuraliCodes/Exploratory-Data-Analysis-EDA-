```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"]=(12,6)
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
df.head()
```

Shape and Info

```python
print("Shape of Dataset:", df.shape)
print("\n Data Info:")
df.info()
```

```python
print("\n Statistical Summary:")
print(df.describe(include='all'))
```

```python
print("\n Missing Values:")
print(df.isnull().sum())
```

Value Counts (Categorical Features)

```python
print("\n Survived Value Counts:\n",df['Survived'].value_counts())
print("\n Pclass Value Counts:\n",df['Pclass'].value_counts())
print("\n Sex Value Counts:\n",df['Sex'].value_counts())
print("\n Embarked Value Counts:\n",df['Embarked'].value_counts())
```

Visualizations

```python
sns.pairplot(df[['Survived','Pclass','Age','SibSp','Parch','Fare']],hue='Survived',palette='Set1')
plt.suptitle("Pairplot-Survival & Features",y=1.02)
plt.show()
```

Heatmap-Correlation

```python
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

Histograms

```python
df.hist(bins=30,figsize=(15,10),color='skyblue')
plt.suptitle("Histograms of Numeric Features")
plt.show()
```

Boxplots-Outliers & Spread

```python
sns.boxplot(x='Pclass',y='Fare',data=df,palette='Pastel1')
plt.title("Boxplot:Fare by class")
plt.show()

sns.boxplot(x='Survived',y='Age',data=df,palette='Set2')
plt.title("Boxplot:Age by Survival")
plt.show()
```

Count Plots -Categorical Features

```python
sns.countplot(x='Survived',data=df,palette='Set1')
plt.title("Count Plot:Survival")
plt.show()

sns.countplot(x='Pclass',data=df,palette='Set1')
```

Summary of Findings

```python
Summary ="""
Most passengers were in Class3 and many of them did not survive.
Females had a significantly higher survival rate than males.
Class1 passengers paid higher fares and had better survival chances.
Younger passengers (children) were more likely to survive.
Some features like 'Fare','Age' and 'Pclass' show visible correlation.
Missing values in 'Age', and 'Cabin' columns need to be handled.
"""
print(Summary)
```
